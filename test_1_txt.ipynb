```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
car_df=pd.read_csv('/mtcars.xls')
print(car_df)
```

```python
a=plt.scatter(car_df['mpg'],car_df['wt'],marker='*')
print(a)
plt.xlabel('mpg')
plt.ylabel('wt')
plt.title('relationship between the weight of the car and mpg')
plt.show()
```

```python
trans=car_df['am'].value_counts()
print(trans)
trans.plot.bar()
plt.xlabel('transmission')
plt.title('thebarplottocheckthefrequencydistributionoftransmissiontype')
plt.show()
```

```python
car_df['mpg'].plot(kind='box')
plt.xlabel('mpg')
plt.title('boxandwhiskerplotformpg')
plt.show()
```

```python
minvalue=np.min(car_df['mpg'])
q1=np.percentile(car_df['mpg'],25)
median=np.median(car_df['mpg'])
q3=np.percentile(car_df['mpg'],75)
maxvalue=np.max(car_df['mpg'])
print('minmum value:',minvalue)
print('Q1:',q1)
print('median:',median)
print('Q3:',q3)
print('maximum values:',maxvalue)
```
